# FinBERT 打分（在 Google Colab 的 GPU 上运行）

本地电脑没有 NVIDIA 显卡，用 CPU 给约 5 万条标题打分要 3 小时左右；Colab 的免费 GPU 只要几分钟。
这个 notebook **只做一件事**：用仓库里同一份 `src/data_utils.py` 和 `src/finbert.py` 生成 `finbert_scores.csv`，再下载回本地 `data/` 文件夹。主分析 `main.ipynb` 会直接读取这个缓存。

步骤：
1. 菜单 **代码执行程序 → 更改运行时类型 → T4 GPU**
2. 运行第一个单元格，一次选中上传 3 个文件：`Combined_News_DJIA.csv`、`src/data_utils.py`、`src/finbert.py`
3. 运行剩下的单元格，浏览器会自动下载 `finbert_scores.csv`，把它放进项目的 `data/` 文件夹

```
import os, shutil
from google.colab import files

uploaded = files.upload()
os.makedirs("src", exist_ok=True)
open("src/__init__.py", "w").close()
for name in ["data_utils.py", "finbert.py"]:
    shutil.move(name, f"src/{name}")
print(sorted(os.listdir(".")), sorted(os.listdir("src")))
```
